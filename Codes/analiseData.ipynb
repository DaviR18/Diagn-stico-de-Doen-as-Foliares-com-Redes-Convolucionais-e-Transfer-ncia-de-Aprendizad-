```python
from pathlib import Path
import pandas as pd
from PIL import Image
import os
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
here = os.getcwd()
pasta = Path(os.path.join(here,"..","PlantVillage-Dataset","raw","color"))
pasta2 = Path(os.path.join(here,"..","PlantVillage-Dataset","leaf_grouping","leaf_maps"))
```

```python
SEMENTE = 42
np.random.seed(SEMENTE)
```

```python
dados = []
for imagem in pasta.rglob("*"):
    if imagem.suffix.lower() not in [".jpg", ".jpeg", ".png"]:
        continue
    dados.append({
        "arquivo": str(imagem),
        "classe": imagem.parent.name.lower().replace("_","").replace("-","").replace(" ", "")
    })
```

```python
df = pd.DataFrame(dados)
```

```python

dados = []
nameMaps = ["Apple_Scab","Apple_Frogeye Spot","Cherry_Healthy","Cherry_Powdery Mildew","Citrus_Citrus Greening June","Citrus_Citrus Greening Oct.","Grape_Black Measles Fungus", "Grape_Black Rot Fungus", "Grape_Leaf Blight Fungus", "Pepper_Healthy", "Pepper_Bacterial Spot JR","Pepper_Bac spot NREC", "Potato_Healthy Potato", "Potato_Early Blight Fungus", "Potato_Late Blight Fungus","Raspberry_Sheet1", "Soy_Sheet1","Strawberry_Scorch", "Tomato_Bacterial Spot Bacteria","Tomato_Early Blight Fungus","Tomato_Late Blight Water Mold","Tomato_Leaf Mold Fungus","Tomato_Septoria Leaf Spot Fungus", "Tomato_Spider Mite Damage Insect", "Tomato_Target Spot Bacteria", "Tomato_YLCV Virus"]

nameClass = ["appleapplescab","appleblackrot","cherry(includingsour)healthy","cherry(includingsour)powderymildew", "orangehaunglongbing(citrusgreening)",
"orangehaunglongbing(citrusgreening)", "grapeesca(blackmeasles)", "grapeblackrot", "grapeleafblight(isariopsisleafspot)","pepper,bellhealthy", "pepper,bellbacterialspot","pepper,bellbacterialspot","potatohealthy", "potatoearlyblight", "potatolateblight","raspberryhealthy", "soybeanhealthy", "strawberryleafscorch","tomatobacterialspot","tomatoearlyblight","tomatolateblight","tomatoleafmold","tomatoseptorialeafspot", "tomatospidermitestwospottedspidermite","tomatotargetspot", "tomatotomatoyellowleafcurlvirus"]

nomes = ["João", "Maria", "Pedro", "Ana"]
for n in range(len(nameMaps)):
    print(nameMaps[n],"=",nameClass[n])
```

```python
indice = 0
dados= []
for arquivo in pasta2.glob("*.csv"):
    tabela = pd.read_csv(arquivo)
    nome = arquivo.stem.lower().replace("_","").replace(" ","")
    if arquivo.stem in nameMaps:
        indice = nameMaps.index(arquivo.stem)
        nome = nameClass[indice]
    
    tabela["class"] =  nome
    tabela["id"] = (nome)+"_"+tabela["Leaf #"].astype(str)
    dados.append(tabela)
```

```python
leaf_df = pd.concat(dados,ignore_index=True)
```

```python
filtro = (
    (leaf_df["class"] == "tomatotargetspot")
    & leaf_df["File Name"].str.startswith("Com.G_FL ", na=False)
)

leaf_df.loc[filtro, "File Name"] = (
    leaf_df.loc[filtro, "File Name"]
    .str.replace("Com.G_FL ", "Com.G_TgS_FL ", regex=False)
)
```

```python
leaf_df
```

```python
print(leaf_df["class"].value_counts().sort_index())
print("Total de classes:", leaf_df["class"].nunique())
```

```python
df['id'] = None
```

```python
for n in range(len(leaf_df)):
    df.loc[df["arquivo"].str.endswith(leaf_df["File Name"][n], na = False)
            & (df["classe"] == leaf_df["class"].iloc[n]),
       "id"] = leaf_df["id"].iloc[n]
```

```python
print("Imagens identificadas:", df["id"].notna().sum())
print("Imagens sem ID:", df["id"].isna().sum())
```

```python
df = df.drop(columns=["id"])
```

```python
treino, restante = train_test_split(df[["arquivo","classe"]],test_size= 0.3, stratify=df["classe"],random_state=SEMENTE)

validacao, teste = train_test_split(restante, test_size=0.5,stratify=restante["classe"], random_state=SEMENTE)
```
